```python
import itertools
import numpy as np
```

```python
scale1 = np.linspace(10, 100, 2 + 1)
scale2 = np.linspace(20, 200, 2 + 1)
scales = np.vstack((scale1, scale2))
scales
```

```python
scales_list = [dim_scales for dim_scales in scales]
scales_list
```

```python
uniformity_values = np.zeros(len(list(itertools.product(*scales_list))))
```

```python
uniformity_values.shape
```

```python
for idx, scale_combo in enumerate(itertools.product(*scales_list)):
    uniformity_values[idx] = float(f"{int(scale_combo[0])}{int(scale_combo[1])}")
```

```python
# Mapeia cada valor de uniformidade ao par de escalas correspondente
# (Útil para saber de qual combinação de escalas cada valor veio)
import pandas as pd

# Produto cartesiano nas mesmas ordem e convenção usadas para preencher uniformity_values
combos = list(itertools.product(*scales_list))  # direita varia mais rápido

# Tabela de mapeamento: uma linha por combinação de escalas
col_names = [f"scale_{i+1}" for i in range(len(scales_list))]
df_mapping = pd.DataFrame(combos, columns=col_names)
df_mapping["uniformity"] = uniformity_values

# Visualiza o mapeamento
df_mapping
```
