```python
from src import Processor
from src.utils import find_doc
```

```python
processor = Processor(index_dir="./index", queries_path="./data/queries.txt", ranker='TFIDF',)
```

```python
query = 'newton law'
results = processor.query(query, "BM25", top_k=45, is_conjuctive=False)
```

```python
results_with_docs = []
for docid, score in results[:43]:
    doc = find_doc('data/corpus.jsonl', docid)
    doc['score'] = float(score)
    results_with_docs.append(doc)
```
