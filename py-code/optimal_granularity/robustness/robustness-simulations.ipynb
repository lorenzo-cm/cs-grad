```python
shapefile_path = "../data/burglaries/burglary.shp"

coords, gdf_coords = load_shapefile_points(shapefile_path)
```

```python
def plot_robustness_scores(robustness_scores: list[float], scales: list[float]):
    plt.figure(figsize=(10, 6))
    sns.lineplot(x=scales, y=robustness_scores)
    plt.xlabel('Scale')
    plt.ylabel('Robustness Score')
    plt.title('Robustness Scores vs. Scale')
    plt.show()
```

```python
size = 15
```

```python
bbox = BoundingBoxBase.from_points(coords)
scales = bbox.gen_scales_from_bbox(size=size,
                                   num_points=len(coords),
                                   min_points_per_square=0.1,
                                   max_percentage_points_per_square=20,
                                   scales_distribution="slow_start_exponential",
                                   method="density")

robust_scores_histogram = robustness(
    points=coords,
    scales=scales,
    bbox=bbox,
    num_simulations=200,
    num_bootstrap=50,
    binning_method="auto",
)

print(robust_scores_histogram)
plot_robustness_scores(robust_scores_histogram, scales[0])
```

```python
bbox = BoundingBoxBase.from_points(coords)
scales = bbox.gen_scales_from_bbox(size=size,
                                   num_points=len(coords),
                                   min_points_per_square=0.1,
                                   max_percentage_points_per_square=20,
                                   scales_distribution="slow_start_exponential",
                                   method="density")

robust_scores_unique = robustness(
    points=coords,
    scales=scales,
    bbox=bbox,
    num_simulations=200,
    num_bootstrap=50,
    binning_method="unique",
)

print(robust_scores_unique)
print(len(robust_scores_unique))

plot_robustness_scores(robust_scores_unique, scales[0])
```
